 Business Dataset Creation</h2>

```python
dates = pd.date_range(start="2025-01-01", periods=12, freq="MS")
data = {
    "Date": dates,
    "Region": ["North", "South", "East", "West"] * 3,
    "Product": ["Electronics", "Fashion", "Home", "Beauty"] * 3,
    "Sales": [125000, 98000, 67000, 54000, 138000, 105000, 72000, 59000,
              152000, 112000, 81000, 63000],
    "Profit": [31000, 22000, 15000, 12000, 35000, 24000, 17000, 13500,
               39000, 26000, 19000, 14500],
    "Orders": [890, 720, 540, 410, 950, 780, 590, 450, 1020, 830, 650, 480]
}
df = pd.DataFrame(data)
df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.strftime("%b %Y")
```

Document Preparation for RAG</h2>

```python
raw_documents = [
    "Q1 2025 Business Report: Revenue reached 1.5 million dollars, up 12 percent year-over-year. North region led sales with strong Electronics demand. Profit margin stood at 22 percent. Supply chain delays affected 5 percent of orders.",
    "Q2 2025 Business Report: Sales grew to 1.8 million dollars with 15 percent increase. Electronics contributed 45 percent of total revenue. South region recorded 25 percent growth thanks to new marketing campaign. Profit reached 450 thousand dollars.",
    "Annual Summary 2024: Total revenue 6.2 million dollars. Asia market grew 30 percent. Customer satisfaction score 4.8 out of 5. Key recommendation is to expand digital channels and optimize inventory management.",
    "Competitor Analysis: Competitor X gained 8 percent market share in Q1. Strategy focus should be on innovation for Product B to regain leadership position."
]

documents = [Document(page_content=doc) for doc in raw_documents]

text_splitter = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=50)
splits = text_splitter.split_documents(documents)
```

<h2 style="
text-align:center;
color:#0d47a1;
font-weight:bold;
font-size:30px;
text-shadow:2px 2px 10px rgba(13,71,161,0.4);
letter-spacing:1px;">
Embeddings and Vector Store Setup</h2>

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={'device': 'cpu'}
)

vectorstore = FAISS.from_documents(splits, embeddings)
print("Vector store created successfully using CPU.")
```

<h2 style="
text-align:center;
color:#0d47a1;
font-weight:bold;
font-size:30px;
text-shadow:2px 2px 10px rgba(13,71,161,0.4);
letter-spacing:1px;">
LLM and RAG Chain Setup</h2>

```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = ""   
os.environ["TOKENIZERS_PARALLELISM"] = "false"

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("gpt2")

pipe = pipeline(
    "text-generation", 
    model=model, 
    tokenizer=tokenizer, 
    max_new_tokens=200, 
    temperature=0.7, 
    do_sample=True,
    device=-1,                    
    device_map=None
)

llm = HuggingFacePipeline(pipeline=pipe)

from langchain_classic.chains import RetrievalQA

qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=vectorstore.as_retriever(search_kwargs={"k": 3}),
    return_source_documents=True
)

print("RAG Chain created successfully with GPU completely disabled.")
```

<h2 style="
text-align:center;
color:#0d47a1;
font-weight:bold;
font-size:30px;
text-shadow:2px 2px 10px rgba(13,71,161,0.4);
letter-spacing:1px;">
Final Example: Testing the RAG Business Analyst Model</h2>

```python
query = "Provide a detailed business performance analysis for Q1 and Q2 2025 including key growth drivers and recommendations."

result = qa_chain.invoke({"query": query})

display(HTML(f'<h2 style="color:#2E7D32;">Query</h2>'))
display(HTML(f'<p style="font-size:16px;color:#263238;">{query}</p>'))

display(HTML(f'<h3 style="color:#1976D2;">Retrieved Context (Top 3 chunks)</h3>'))
for i, doc in enumerate(result["source_documents"], 1):
    display(HTML(f'<div style="background:#E3F2FD;padding:12px;margin:8px 0;border-radius:8px;"><strong>Chunk {i}:</strong> {doc.page_content}</div>'))

display(HTML(f'<h3 style="color:#8E24AA;">Generated Analysis by RAG Model</h3>'))
display(HTML(f'<div style="background:#F3E5F5;padding:15px;border-radius:8px;font-size:15.5px;line-height:1.6;">{result["result"]}</div>'))

display(HTML('<h4 style="color:#2E7D32;"> This colored output demonstrates the full strength of the RAG model: accurate retrieval + contextual generation.</h4>'))
```
